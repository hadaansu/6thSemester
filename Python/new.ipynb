```python
import pandas as pd
import numpy as np
df=pd.read_csv("marksheet.csv")
df
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Section']=le.fit_transform(df['Section'])
df["Gender"]=le.fit_transform(df["Gender"])
df
```

```python
#select features (X) and target (Y)
X=df[["Gender","Age","Section"]]
Y=df["Maths"]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
#TRAIN ON 80% DATA
#TEST ON 20% DATA
```

```python
#train model
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,Y_train)
```

```python
#predict model
y_pred=model.predict(X_test)
print(y_pred[:5])
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
#evaluate model
from sklearn.metrics import mean_absolute_error, mean_squared_error,r2_score
mae=mean_absolute_error(Y_test,y_pred)
mse=mean_squared_error(Y_test,y_pred)
r2=r2_score(Y_test,y_pred)
print("Mean Absolute Error:",mae)
print("Mean Squared Error:",mse)
print("R2 Score:",r2)
```

```python
#mae :averages of absolute errors
#mse : square of errors
#r2: 1- perfect
   #0- bad
   #negative-very bad
#r2=0.85 ---->model explain 85% data
#r2=0.30 ----->model explain 30% data(weak model)
#mae ---> average mistake (how far off)
#mse ----> penalizes big wrong prediction
#r2 ----> how good overall model is 

```

```python
#Regression model is used to predicting numerical values based on input features.
#Linear regression is a simple and widely used regression model that tries to find the best-fitting straight line 
# (linear relationship) between the input features and the target variable. 
#y=mx+c

#1) Linear Regression
#2) Random Forest Regression

# A straight line might not represent relationship very well so for that purpose random forest can be useful.
#in random forest many different trees works together.
```

```python
#Linear regression                      random forest
# find best-fit line                    combine many decision trees
#mainly linear                          can learn non-linear pattern
#                                       different rules and interactions
```

```python
#   ACTUAL         PREDICTED           ERROR(MAE)          ERROR(MSE)
#   80              78                  2                    2^2=4
#   70              68                  2                    2^2=4
#   90              50                  40                   40^2=1600

#MAE TREATS ALL ERROR EQUALLY:
# MAE+(2+2+40)/3 =14.6

#MSE: square of errors
#  #MSE: (2^2+2^2+40^2)/3 =536

#MAE= # mae = how far are my predictions from the actual marks on average?
# mse = are there any very large mistakes that i should punish heavily
# mse= gives much more weights to large errors 

```

```python
# train model
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier()
model.fit(X_train,Y_train)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
#predict model
y_pred = model.predict(X_test)
print(y_pred[:5])
```

```python
#evaluate model
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

mae=mean_absolute_error(Y_test,y_pred)
mse=mean_squared_error(Y_test,y_pred)
r2=r2_score(Y_test,y_pred)

print("MAE : ",mae)
print("MSE : ",mse)
print("R2_score : ",r2)
```
