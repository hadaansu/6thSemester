```python
import pandas as pd
import numpy as np
df = pd.read_csv("marksheet.csv")
```

```python
import matplotlib.pyplot as plt

subjects = ["Science", "English", "History", "Maths"]
marks = df.loc[0, subjects]      # First student

plt.figure(figsize=(6,4))
plt.bar(subjects, marks)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Marks of First Student")

plt.show()
```

```python
subjects = ["Science", "English", "History", "Maths"]
marks = df.loc[0, subjects] 
plt.figure(figsize=(11,7))

plt.plot(subjects,marks,marker='o')

plt.xlabel("Subject")
plt.ylabel("marks")

plt.title("Marks trend over subjects")

plt.show()
```

```python
subjects = ["Science", "English", "History", "Maths"]
marks = df.loc[0, subjects] 
plt.figure(figsize=(11,7))

plt.plot(subjects,marks,marker='^')

plt.xlabel("Subjects")
plt.ylabel("marks")

plt.title("Marks trend over subjects")

plt.show()
```

```python
subjects = ["Science", "English", "History", "Maths"]
marks = df.loc[0, subjects] 


plt.figure(figsize=(5,5))

plt.scatter(subjects,marks,label=" ")

plt.xlabel("Subject")
plt.ylabel("Marks")

plt.title("Scatter plot : Subject vs marks ")

plt.legend()
# plt.grid()
plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.boxplot([df["Science"]])

plt.xlabel("Science")
plt.ylabel("Marks")
plt.title("Box Plot")

plt.show()
```

```python
#select features (X) and target(Y)
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv("college_sleep_and_gpa.csv")
X = df['gender']
Y = df['avg_sleep_minutes']

#Boxplot
plt.boxplot(Y)
plt.ylabel("Average Sleep Minutes")
plt.title("Boxplot")
plt.show()
#The box in the plot represents the middle 50% of students’ sleep time, which is roughly between 370 and 430 minutes,
#  while the line inside the box shows the median at around 400 minutes (about 6.5–7 hours), meaning half the students sleep 
# less than this and half sleep more. The whiskers extend to about 270 minutes on the lower side and 520 minutes on the upper side,
#  indicating the normal range of sleep duration, and the small dots beyond these whiskers represent outliers, showing some
#  students sleep very little (around 190–260 minutes) and others sleep much more (around 550–590 minutes).


```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("college_sleep_and_gpa.csv")

# Numeric columns
X = df['avg_sleep_minutes']
Y = df['term_gpa']

# Scatter plot
plt.scatter(X, Y)
plt.xlabel("Average Sleep Minutes")
plt.ylabel("Term GPA")
plt.title("Sleep vs GPA Scatterplot")

plt.show()
#Most students cluster around 350–450 minutes of sleep and GPA between 3.0 and 4.0, 
# showing that many students perform well regardless of small differences in sleep. T
```

```python
#Scatterplot
plt.scatter(X['Science'], Y)
plt.xlabel("Science")
plt.ylabel("Maths")
plt.title("Scatterplot")
plt.show()
#The dot represents a student. X being Marks in science and Y being marks in maths.
#There appears to be no correlation between marks in science and maths as the dots are scattered all over the place.

#Line graph
plt.plot(Y)
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.title("Line Graph")
plt.show()
#In the line graph, x-axis shows all studnets(1-250) and y axis shows marks in maths.
#Since the students are not arranged in any order, the line graph appears to be random ups and downs


#Bargraph
plt.bar(df.index, Y)
plt.xlabel("Student")
plt.ylabel("Maths")
plt.title("Bar Graph")
plt.show()
#The bar graph shows the marks of each student in the class.
#Each bar represents a student and the height of the bar indicates their marks in maths.

#Histogram
plt.hist(Y)
plt.xlabel("Maths")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
#The histogram shows the distribution of marks in maths
# The x-axis represents the range of marks and the y-axis represents the frequency of students
# scoring within that range. 
# It helps to visualize how many students scored
#  in different intervals of marks.
```
