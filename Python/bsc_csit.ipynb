```python
expenses = []

def add_expense():
    name = input("Enter expense name: ")
    amount = float(input("Enter amount: "))

    expenses.append([name, amount])
    print("Expense added!")


def view_expense():
    if len(expenses) == 0:
        print("No expenses found.")
    else:
        print("\nExpenses:")
        for e in expenses:
            print(e[0], "-", e[1])


def show_total():
    total = 0

    for e in expenses:
        total = total + e[1]

    print("Total Expense =", total)


while True:
    print("\nExpense Tracker")
    print("1. Add Expense")
    print("2. View Expenses")
    print("3. Show Total")
    print("4. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        add_expense()

    elif choice == "2":
        view_expense()

    elif choice == "3":
        show_total()

    elif choice == "4":
        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```

```python
#question: Bank Account System
#create a program to manage bank account.
#acccount holder name
#types of account
#balance

#the program should:
#create multiple accounts
#deposit money
#withdraw money
#display account details
class BankAccount:
    def __init__(self, account_holder, account_type, initial_balance=0):
        self.account_holder = account_holder
        self.account_type = account_type
        self.balance = initial_balance
    
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            return f"Deposited Rs. {amount}. New balance: Rs. {self.balance}"
        return "Amount must be positive"
    
    def withdraw(self, amount):
        if amount > 0:
            if amount <= self.balance:
                self.balance -= amount
                return f"Withdrew Rs. {amount}. New balance: Rs. {self.balance}"
            else:
                return f"Insufficient balance! You have Rs. {self.balance}"
        return "Amount must be positive"
    
    def display(self):
        return f"Holder: {self.account_holder}, Type: {self.account_type}, Balance: Rs. {self.balance}"

acc1 = BankAccount("Anshu", "Savings", 50000)
acc2 = BankAccount("Alice", "Current", 100000)

print(acc1.deposit(10000))
print(acc1.withdraw(5000))
print(acc1.display())
```

```python

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")

class Person(Student):
    def __init__(self, name, age, roll_no):
        super().__init__(name, age)   # Calls parent constructor
        self.roll_no = roll_no

    def show_student(self):
        print(f"Roll No: {self.roll_no}")

# Creating object of child class
s1 = Person("Ram", 32, 1)

s1.display()
s1.show_student()
```

```python
#Ecommerce
class Product:
     def __init__(self, name, price):
          self.nam=name
          self.pric=price
     def display(self):
             print(f"Name: {self.nam}")
             print(f"Price: {self.pric}")
class electronics(Product):
     def __init__(self,name,price,warranty):
          super().__init__(name, price)  
          self.warr = warranty
     def show(self):
        super().display()
        print(f"Warranty: {self.warr}")
class clothing(Product):
     def __init__(self,name,price,size):
          super().__init__(name, price)  
          self.size=size
     def sow(self):
          super().display()
          print(f"Size:{self.size}")
s1= electronics("Mobile",7000,2)
s2= clothing("Pant",500,1)
s1.show()
s2.sow()
    
```

```python
#polymorphism =many forms
#the same method name behaves differently for different objects
#like: one function --->different behaviour

#real life examples:
#notification system
# (facebook,insta,tiktok)
#sms notification
#email notification
#all used to send notification

```

```python
class Notification:
    def send (self):
        print("Sending notification...")
class Email:
    def send(self):
        print("Sending emails")
class Sms(Notification):
    def send(self):
        print("sending sms")
class Push(Notification):
    def send(self):
        print("push notification")
notification=[Email(),Sms(),Push()]
for n in notification:
    n.send()
```

```python
#same name --> different behavior
#create a shape system:
#base class: shape
#child class:circle,rectangle
#method:area() --->behaves differently

class shape:
    def area(self):
        pass
class circle(shape):
    def __init__(self,r):
        self.r=r
    def area(self):
        return 3.14* self.r*self.r
class rectangle(shape):
    def __init__(self,l,b):
        self.l=l
        self.b=b
    def area(self):
        return self.l*self.b
shapes=[circle(5),rectangle(8,9)]
for s in shapes:
    print(s.area())
```

```python
#without using super whether we can call the parent constructor or not?

#Yes,we can

#Using super
class Parent:
    def __init__(self):
        print("Parent constructor")

class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child constructor")

obj = Child()

#without using super
class Parent:
    def __init__(self):
        print("Parent constructor")

class Child(Parent):
    def __init__(self):
        Parent.__init__(self)
        print("Child constructor")

obj = Child()


#1) when you dont need parents __init__() ---> no need of super class
#2) child has its own __init--() and needs parent attributes ---> use super class
```

```python
#types of polymorphism
# design library system using inheritance
#base class: libraryitems
#derivedclass: book
#library item: title,author,category,price
#methods display info,apply discount,final price
```

```python
class LibraryItem:
    def __init__(self, title, author, category, price):
        self.title = title
        self.author = author
        self.category = category
        self.price = price
    def display_info(self):
                print(f"Title: {self.title}")
                print(f"Author: {self.author}")
                print(f"Category: {self.category}")
                print(f"Price: {self.price}")
    def apply_discount(self, discount):
        self.price = self.price - (self.price * discount / 100)

    def pricee(self):
        print("Final Price:", self.price)
class Book(LibraryItem):
    def __init__(self, title, author, category, price, pages):
        super().__init__(title, author, category, price)
        self.pages = pages

    def display_info(self):
        super().display_info()
        print("Pages:", self.pages)

book1 = Book("Python", "priya", "programming", 1000, 500)
book1.display_info()
book1.apply_discount(20)
book1.pricee()

```

```python
#Polymorphism means "one method, many forms."
#Types of polymorphism

#Compile-time Polymorphism :   (Method Overloading)
#             The same method is used with different parameters.
class Calculator:
    def add(self, a, b=0):
        print(a + b)
c = Calculator()

c.add(5)
c.add(5, 10)

#Runtime Polymorphism:   (Method Overriding)
#              A child class provides its own version of a method already defined in the parent class.
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")

d = Dog()
c = Cat()
d.sound()
c.sound()
```

```python
#overriding
#parents has a method ---> childs inherits it --> child changes the same method --->method overiding


#use super with overriding
class Person:
    def show(self):
        print("I am person")
class student(Person):
    def show(self):
        super().show()
        print('i am student')
s=student()
s.show()
```

```python
# 2 ta parents line ani child le 2tai ko inherit garne paryo
#ATM system
# Account(Parent)
# SavingAccount(child)
#Data types:
# public type---> account holder name
# protected--->_balance
# private ---->__pin
# mthods: deposits,withdeaw ,show_balance
```

```python
class Account:
    def __init__(self,name,balance,pin):
        self.name=name
        self._balance=balance #protected
        self.__pin=pin #private
    def deposit(self,amount):
        self._balance+=amount
    def withdraw(self,amount):
        self._balance-=amount
    def show_balance(self):
        print("Total balance=",self._balance)
acc=Account("Anshu",50000,100)
acc.deposit(1000)
acc.withdraw(5000)
acc.show_balance()
```

```python
# design hospital system where
# base class : Treatment(abstract)
# child class : surgery, therapy

#each must implement
#perform_treatment()

from abc import ABC , abstractmethod

class Treatment(ABC):
    @abstractmethod
    def perform_treatment(self):
        pass

class Surgery(Treatment):
    def perform_treatment(self):
        print("Performing surgery with doctors ")

class Therapy (Treatment):

    def perform_treatment(self):
        print("Providing therapy sessions to patient")

#objects:
t1=Surgery()
t2=Therapy()
t1.perform_treatment()
t2.perform_treatment()
```

```python
#decoraor: modifies or add behavior to a function without changing function's code
def greet():
    print ("hello")
greet() 
```

```python
def my_decorator(func):
    def wrapper():
        print("before")
        func()
        print("After")
    return wrapper
```

```python
@my_decorator
def greet():
    print ("hello")
greet() 
```

```python
from abc import ABC , abstractmethod
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width
r = Rectangle(5, 4)
print("Rectangle Area:", r.area())
```

```python
# 2 ta parents line ani child le 2tai ko inherit garna paryo
class Person:
    def __init__(self,name,id):
        self.name=name
        self.id=id
    def show(self):
        print(f"Name : {self.name}\nID : {self.id}")
class Employee():
    def __init__(self,salary):
        self.salary=salary
    def show(self):
        print(f"Salary : {self.salary}")
class Manager(Person,Employee):
    def __init__(self,name,id,salary,experience):
        Person.__init__(self,name,id)
        Employee.__init__(self,salary)
        self.experience=experience
    def show(self):
        Person.show(self)
        Employee.show(self)
        print(f"Experience : {self.experience} yrs")

M = Manager("ram",101,50000,3)
M.show()
```
